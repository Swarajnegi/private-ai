## 1. Iteration & Tuple Unpacking

### Concepts:
- **`enumerate(iterable, start=0)`**: Yields pairs of `(index, element)`.
- **`zip(*iterables)`**: Aggregates elements from each of the iterables in parallel.
- **`dict.items()`**: Yields `(key, value)` pairs. Iterating a dict directly (`for k in d:`) yields **only keys**.

### The Pitfall:
```python

    ...
```

## 2. Sorting Contracts & The `key=` Parameter

### Concepts:
- **`list.sort(key=..., reverse=...)`**: **In-place mutation**. Modifies the list and returns `None`. Only works on lists.
- **`sorted(iterable, key=..., reverse=...)`**: **Pure function**. Accepts any iterable and returns a **new sorted list**.
- **Tuple Sorting**: By default, tuples are sorted lexicographically by index 0, then index 1, etc.
- **Custom Sort Key**: Use `key=lambda x: x[1]` to sort tuples/objects by a specific attribute.

### The Pitfall:
```python
nums = [(0, 7), (1, 2), (2, 5)]
res = nums.sort()   # res is None!
nums.sort()         # Sorts by index (0, 1, 2), NOT by value (7, 2, 5)!
```
